# PySpark via Enterprise Gateway

**Use the PySpark / spark_python_operator kernel.** That kernel runs on Enterprise Gateway and creates a SparkApplication (driver + executors) via the Spark Operator.

Run the cells below; the first run will start the SparkApplication with **3 executors**. Watch the namespace for the driver and executor pods.

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .config("spark.executor.instances", "3")
    .getOrCreate()
)
df = spark.range(10)
df.show()
```

```python
# Same kernel (same SparkApplication — 1 driver + 3 executor pods in the namespace)
spark.sql("SELECT 1 as one").show()
```
